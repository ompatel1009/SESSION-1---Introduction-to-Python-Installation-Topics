```python
def total_cart_amount(prices):
    return sum(float(price) for price in prices)

# Sample Flipkart-style cart
cart_prices = ['1299.00', '499.50', '2999', '150.75', '89.99']
total = total_cart_amount(cart_prices)

print(f'Cart items: {cart_prices}')
print(f'Total Cart Amount: ₹{total}')
```
