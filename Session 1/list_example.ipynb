```python
""" 
Collection : collection is a most powerfull data structure in python 

    there are mainly 4 types of collection 

    1) list 
    2) tuple 
    3) set 
    4) dictionary 

1) list : list a most powerfull concept of collection . 

    list a structure data type which is represent by [] braces. 
    list which is contain similar and dis-similar data elements. 

    list is a oderable , sequenceble , indexable , and mutable data type. 

    mutable : we can change it value. 
""" 
```

```python
""" 
how to create list : 

there are 2 ways to create list 

1) using of square braces : 

    e.g.    l1 = [] 

2) using of list constructor 

    e.g.    l1 = list() 


""" 
l1 = []  # blank list 
print(l1)
print(type(l1))

l2 = list()
print(l2)
print(type(l2))
```

```python
l1 = [23,45,67,98,61,67]

print(l1)

# access list using loop 

for i in l1:
    print(i)
```

```python
l1 = [12,4,67,8,56]

print(max(l1))
print(min(l1))
print(sum(l1))

l1.reverse()
print(l1)

l1.sort() 
print(l1)

l1.sort(reverse=True)
print(l1)
```
